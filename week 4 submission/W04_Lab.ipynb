### D1. Create a list and index into it

1. Create `labels = ["R-001", "R-002", "R-003"]`.
2. Print the first item with `labels[0]`, the last item with a **negative** index, and the length with `len()`.

**Expected output:** `R-001`, `R-003`, `3`

````python
# D1
"""Access items in a list using index"""
labels = ["R-001","R-002","R-003"]
print(labels[0]) #list has index 0 = position 1
print(labels[-1]) # the last item =index -1
print(len(labels)) #prints 3 as the list contains 3 items
````

### D2. Slicing

1. Create `values = [5, 10, 15, 20, 25, 30]`.
2. Use slicing to print: the middle four, the first two, and everything from position 3 to the end.

**Expected output:**
```
[10, 15, 20, 25]
[5, 10]
[20, 25, 30]
```

*Remember: the stop position is not included.*

````python
# D2
"""Slice through a list of numbers"""
values = [5, 10, 15, 20, 25, 30]
print(values[1:5]) # starts with index 1 and stops at index 4,index  5 is exclusive
print(values[:2]) # by default, starts at index 0 and stops at index 1, index 2 is exclusive
print(values[2:]) # position 3 is index 2. It starts at index 2 and ends with the last item

````

### D3. Change a list, then loop over it

### D5. Strings index and slice the same way

### D6. List comprehensions

1. Create `values = [40, 55, 70, 25]`.
2. Using a list comprehension, make a new list with every value doubled, and print it.
3. Using a list comprehension with `if`, make a new list with only the values that are 50 or more, and print it.

**Expected output:**
```
[80, 110, 140, 50]
[55, 70]
```

````python
# D6
"""Create a list from another list"""
values = [40, 55, 70, 25]
doubled = [value* 2 for value in values] # doubles the value from the list 'values' and store them in a new list
print(doubled)
over_50 = [value for value in values if value >= 50] # check which items in values >= 50 and store them in a new list
print(over_50)
````

*Walkthrough: `02 - Dictionaries`*

### D7. Create, read, add and update

1. Create `record = {"label": "R-001", "value": 87.0, "limit": 100.0}`.
2. Print the label using its key.
3. Add a new key `"status"` with the value `"OK"`.
4. Change `"value"` to `90.0`.
5. Print the whole dict.

**Expected output:**
```
R-001
{'label': 'R-001', 'value': 90.0, 'limit': 100.0, 'status': 'OK'}
```

````python
# D7
"""Update a dictionary by changing values and adding items"""
record = {"label": "R-001", "value": 87.0, "limit":100.0}
print(record["label"]) # prints the value stored in the key label
record["status"]="OK" # adding a new key 'status' and its value
record["value"] = 90.0 # changing the value from 87.0 to 90.0
print(record)
````

### D8. Missing keys and `.get()`

Using `record` from D7:

1. Print `record.get("status")`.
2. Print `record.get("owner")`: this key does not exist.
3. Print `record.get("owner", "n/a")`.

**Expected output:** `OK`, `None`, `n/a`

````python
# D8
"""Use .get() to return none or a default to a missing keys without raising an error."""
record = {"label": "R-001", "value": 87.0, "limit":100.0,"status":"OK"}
print(record.get("status")) # prints the value of key 'status'
print(record.get("owner")) # prints None even if key does not exist as .get() avoids an error from raising
print(record.get("owner", "n/a")) # prints n/a as a default value of the nonexistent key 'owner'
````

### D9. Loop over a dict

Using `record` from D7, write three loops:

a) print only the **keys** (`for key in record:`)

b) print only the **values** (`.values()`)

c) print both, as `key: value` (`.items()`)

**Expected output for c):**
```
label: R-001
value: 90.0
limit: 100.0
status: OK
```

````python
# D9
"""Access the keys only, the values only then both"""
record = {"label": "R-001", "value": 87.0, "limit":100.0, "status":"OK"}
for key in record:
    print(key) # prints only the keys to a dictionary
print() # add a space for neater output
print(record.values()) # prints the values only
print()
for key,values in record.items():
    print(f"{key}:{values}") # prints both the keys and the values
````

### D10. Build a dict from two lists with `zip()`

1. Create `hostnames = ["srv-01", "srv-02", "srv-03"]` and `used = [87, 45, 120]`.
2. Build a dict with `dict(zip(hostnames, used))` and print it.
3. Print the value for `"srv-03"`.

**Expected output:**
```
{'srv-01': 87, 'srv-02': 45, 'srv-03': 120}
120
```

````python
# D10
"""Turn 2 lists into a dictionary"""
hostnames = ["srv-01", "srv-02", "srv-03"]
used = [87, 45, 120]
dictionary= dict(zip(hostnames, used)) # zip pairs the 2 lists together while dict() makes them into a dictionary
print(dictionary)
print(dictionary["srv-03"]) # accessing the value to the key 'srv-03' 
````

### D11. Sets: unique values and quick checks

1. Create `ips = ["10.0.0.5", "10.0.0.9", "10.0.0.5", "10.0.0.7"]`.
2. Make a set from the list, and print how many **unique** IPs there are.
3. Print whether `"10.0.0.9"` is in the set, and whether `"10.0.0.1"` is.

**Expected output:** `3`, `True`, `False`

````python
# D11
"""Convert a list to a set"""
ips = ["10.0.0.5", "10.0.0.9", "10.0.0.5", "10.0.0.7"]
new_set=set(ips) # the list is converted to a set, hence duplicates are removed
print(f"There are {len(new_set)} unique IPs") # len(new_set) counts the number of unique items in the set
print("10.0.0.9" in new_set) # ask a question(boolean):is 10.0.0.9 in the set?
print("10.0.0.1" in new_set) # ask a question(boolean):is 10.0.0.9 in the set?
      
````

### D13. Read a file line by line and split each line

`records_sample.txt` (in this `LAB` folder) has one record per line, like `R-001,87,100`.

1. Open it with `with` and loop over it line by line.
2. For each line, use `.strip()`, then `.split(",")` into `label`, `value`, `limit`.
3. Convert `value` and `limit` with `float()`, and print all three.

**Expected output:**
```
R-001 87.0 100.0
R-002 120.0 100.0
R-003 45.0 100.0
```

````python
# D13
"""Read and split a file line by line"""
with open("records_sample.txt") as f: # opens the file record_sample.txt
    for line in f: #looping through each line in the file
        label, value, limit= line.strip().split(",") #removing whitespaces at the beginning and end
                                                     #breaks the line into pieces and assign them to label, value and limit respectively
        value=float(value) # converting value from string to float
        limit=float(limit) ## converting limit from string to float

        print(value, label, limit)
````

*Walkthrough: `04 - Handling Errors`*

### D14. Catch a `ValueError`

1. Create `text = "abc"`.
2. Inside `try:`, convert it with `float(text)` and print the number.
3. In `except ValueError:`, print `Not a number: abc` instead.
4. Now change `text` to `"42"` and run again.

**Expected output:** `Not a number: abc`, then (after the change) `42.0`

````python
# D14
"""Handling a ValueError using try and except"""
text = "42" # storing a string in the variable text
try: # will raise value error if there is(if text is not a valid number)
    text=float(text) #converting text from string to float
    print(text)

except ValueError:
    print("Not a number: abc")#if there ia a ValueError, this is printed instead of crashing the code
    
````

### D15. Catch a `FileNotFoundError`

1. Inside `try:`, open `missing.txt` with `with` and print its contents.
2. In `except FileNotFoundError:`, print `missing.txt not found`.

**Expected output:** `missing.txt not found`

````python
# D15
"""Handle a FileNotFoundError with try and except."""
try:
    with open("missing.txt") as f: # opens the file 'mising.txt' and raises a FileNotFoundError if it does not exist
        content=f.read()
        print(content) # prints the content of the file 'mising.txt'

except FileNotFoundError:
    print("missing.txt not found") # instead of the program crashing, this is printed if the file does not exist 
````

### C1. Load records safely from a file

Write `load_records(path)` with a one-line docstring. It reads `label,value,limit` lines from a file
and returns a **list of dicts**. If the file does not exist, it prints a clear message and returns an
empty list instead of crashing.

Then load `records_sample.txt` and print each record's label and status, as in D16.

**Test it:**

| Call | Expected output |
|---|---|
| `load_records("records_sample.txt")` | the same three lines as D16 |
| `load_records("nope.txt")` | a clear message, then `[]` |

````python
# C1
def load_records(path):
    """Read value, limit and label and returns list of dicts"""
    list_of_dicts = []
    try:
      with open(path) as f:
         for line in f:
            label, value, limit = line.strip().split(",")
            list_of_dicts.append({"label": label,
                                  "value":float(value),
                                  "limit": float(limit),})
      return list_of_dicts
    except FileNotFoundError:
       print("This file does not exist")
       return [ ]
records = load_records("records_sample.txt")      
for items in records: # loops through each dictionary
    if items["value"] >= items["limit"]: # if value>= limit, it print over limit
        print(f"{items["label"]}: OVER LIMIT")
    else: # if value <= limit, it prints ok
        print(f"{items["label"]}: OK")
````

### C2. Write the results to a file

Using your records from C1, write one line per record to a new file called `results.txt`,
in the form `label,status`. Then read `results.txt` back and print it.

**Expected output:**
```
R-001,OK
R-002,OVER LIMIT
R-003,OK
```

````python
# C2

""""Write one line per record in the form of label and status"""
records = [
    {'label': 'R-001', 'value': 87.0, 'limit': 100.0},
    {'label': 'R-002', 'value': 120.0, 'limit': 100.0}, 
    {'label': 'R-003', 'value': 45.0, 'limit': 100.0}]

with open("results.txt","w") as f: # the file is opened for us to write
    for record in records: #loop through every record
        if record["value"] >= record["limit"]: #check if the condition is true
            status = "OVER LIMIT" # assign a value to status based on the condition
        else:
            status = "OK" # assign another value to status if condition was false
        f.write(f"{record['label']}, {status}\n") # writes the label and status

with open("results.txt","r") as f: # the file is opened again for reading
    print(f.read()) # the file's content is read 
````
